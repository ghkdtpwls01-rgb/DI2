# 문제 1. 게임 아이템 뽑기 확률, 정말 10%일까? (난이도: 하)

**📘 문제**

- 어떤 게임 개발사에서 새로운 아이템의 뽑기 확률이 **10%**로 설정되었다고 주장하고 있습니다.

- 그런데 유저 커뮤니티에서는 "실제 확률은 10%보다 낮은 것 같다"는 의혹이 제기되었습니다.

- 이를 확인하기 위해 한 유저가 아이템을 **200번 뽑았고, 그중 12번 성공**했습니다.

- 이번 실습에서는 **이항검정(Binomial Test)**으로 이 결과가 개발사의 주장을 기각할 만큼 유의미한지 확인하고,
  시뮬레이션을 통해 **p-값이 무엇을 의미하는지**를 직접 눈으로 확인해 봅니다.

```python
# 문제 설정
n = 200  # 총 시행 횟수
x = 12   # 관찰된 성공 횟수
p = 0.1  # 귀무가설에서의 성공 확률

print(f"관찰 결과: {n}번 시도 중 {x}번 성공 (관찰 성공률 {x/n:.1%})")
```

**📌 아래를 수행해 보세요:**

- 귀무가설(H₀)과 대립가설(H₁)을 세워 봅시다.

- 이항검정을 수행하여 p-값을 구하고, 유의수준 0.05를 기준으로 결론을 내려봅시다.

- **"만약 개발사의 주장대로 실제 확률이 정말 10%라면"** 200번 뽑기에서 성공 횟수가 어떻게 분포하는지 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 우리가 관찰한 '12번 성공'이 얼마나 희귀한 일인지 관찰해 봅시다.

```python
# [문제 1] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 확률이 10%다(차이가 없다)
# H₁ (대립가설): 확률이 더 낮다(차이가 있다)
```

```python
# [문제 1] Q2. 이항검정을 수행하고 p-값을 계산해봅시다.
# '확률이 더 낮은 것 같다'는 주장을 검증하려면 alternative 인수를 어떻게 설정해야 할까요?

# 여기에 코드를 작성해주세요.
# [문제 1] Q2. 이항검정 수행 및 p-값 계산

# '확률이 더 낮은 것 같다'는 주장 -> H1: p < 0.10
# 이는 분포의 "왼쪽 꼬리(더 적은 성공 횟수 쪽)"만 보는 것이므로
# alternative='less' 로 설정해야 한다.
#   - 'less'      : H1: p < p0  (관찰값이 기대보다 작은 쪽 방향)
#   - 'greater'   : H1: p > p0  (관찰값이 기대보다 큰 쪽 방향)
#   - 'two-sided' : H1: p != p0 (양방향)

result = stats.binomtest(x, n, p, alternative='less')
p_value = result.pvalue

print(f"검정통계량 기반 p-value: {p_value:.5f}")

#[이항검정 결과 (alternative='less')]
#관찰 성공 횟수: 12 / 200
#귀무가설 성공확률(p0): 0.1
#검정통계량 기반 p-value: 0.03205
#=> p-value(0.03205) < 유의수준(0.05) 이므로 H0을 기각한다.
#   즉, 실제 확률은 10%보다 유의하게 낮다고 볼 수 있는 통계적 근거가 있다.
```

```python
# [문제 1] Q3. 유의수준 0.05를 기준으로 통계적 결론을 내리는 코드를 작성해봅시다.
# alpha = 0.05

# 여기에 코드를 작성해주세요.
# [문제 1] Q3. 유의수준 0.05 기준 통계적 결론
alpha = 0.05

print(f"[통계적 결론]")
print(f"p-value = {p_value:.5f}")
print(f"alpha   = {alpha}")

if p_value < alpha:
    conclusion = (
        "귀무가설(H0)을 기각한다.\n"
        "  -> 통계적으로 유의한 수준에서, 실제 아이템 뽑기 확률은\n"
        "     개발사가 주장한 10%보다 낮다고 판단할 수 있다.\n"
        "  -> 유저 커뮤니티의 의혹에 통계적 근거가 있다."
    )
else:
    conclusion = (
        "귀무가설(H0)을 기각할 수 없다.\n"
        "  -> 관찰된 결과(12/200)만으로는 실제 확률이 10%보다\n"
        "     낮다고 볼 만한 충분한 통계적 근거가 없다.\n"
        "  -> 즉, 이번 표본 결과는 우연(표본 변동)의 범위 내에 있을 수 있다."
    )

print(conclusion)
```

```python
# [문제 1] Q4. H₀가 사실(실제 확률 = 10%)이라고 가정하고,
# 200번 뽑기를 10,000번 반복하는 시뮬레이션을 수행해봅시다.
# 힌트: numpy의 이항분포 랜덤 함수(np.random.binomial)를 사용해보세요.

# 여기에 코드를 작성해주세요.
# [문제 1] Q4. H0가 사실이라는 가정 하의 시뮬레이션

# H0: 실제 성공 확률 = 10%
# 이 가정 하에서 "200번 뽑기"라는 실험을 10,000번 반복했을 때
# 성공 횟수가 어떻게 분포하는지를 직접 만들어본다.

np.random.seed(42)  # 재현성을 위한 시드 고정 (교육용, 실무에서는 생략 가능)

n_simulations = 10000
simulated_successes = np.random.binomial(n=n, p=p, size=n_simulations)
```

# 문제 2. 과자 한 봉지의 중량은 150g이 맞을까? (난이도: 하)

**📘 문제**

- 한 식품 공장에서 생산하는 과자 한 봉지의 **목표 중량은 150g**입니다.

- 품질관리팀은 생산 공정이 목표 중량을 잘 맞추고 있는지 확인하기 위해,
  생산된 과자 **30봉지를 무작위로 추출**하여 무게를 측정했습니다.

- 이번 실습에서는 **일표본 t-검정(One-sample t-Test)**으로 표본 평균이 목표 중량과 유의미하게 다른지 검증하고,
  **"평균이 정확히 150g이라면"** 어떤 표본 평균들이 나올 수 있는지 시뮬레이션으로 확인해 봅니다.

```python
# 데이터 생성 (과자 30봉지의 무게 측정값)
np.random.seed(42)
sample_weights = np.random.normal(loc=151.5, scale=2, size=30)
pop_mean = 150  # 목표 중량 (모평균)

# 표본 분포 시각화
sns.histplot(sample_weights, bins=10, kde=True)
plt.axvline(x=pop_mean, color='gray', linestyle='--', label='목표 중량 150g')
plt.title("추출한 과자 30봉지의 무게 분포")
plt.xlabel("무게 (g)")
plt.legend()
plt.show()
```

**📌 아래를 수행해 보세요:**

- 귀무가설(H₀)과 대립가설(H₁)을 세워 봅시다.

- 일표본 t-검정을 수행하여 t-통계량과 p-값을 구하고, 유의수준 0.05를 기준으로 결론을 내려봅시다.

- **"평균 무게가 정확히 150g이라면"** 30개짜리 표본의 평균이 어떻게 분포하는지 10,000번 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 실제 관찰된 표본 평균이 그 분포에서 어디에 위치하는지 관찰해 봅시다.

```python
# [문제 2] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 표본평균 = 모집단 평균 같다
# H₁ (대립가설): 다르다
```

```python
# [문제 2] Q2. 일표본 t-검정을 수행하여 t-통계량과 p-값을 구해봅시다.

# 여기에 코드를 작성해주세요.
# [문제 2] Q2. 일표본 t-검정(One-sample t-test)
from scipy import stats

# H1: mu != 150 (양측검정) -> alternative='two-sided'
t_stat, p_value_t = stats.ttest_1samp(sample_weights, popmean=pop_mean, alternative='two-sided')

print(f"t-통계량: {t_stat:.4f}")
print(f"p-value: {p_value_t:.5f}")
```

```python
# [문제 2] Q3. 유의수준 0.05를 기준으로 통계적 결론을 내리는 코드를 작성해봅시다.
# alpha = 0.05

# 여기에 코드를 작성해주세요.
# [문제 2] Q3. 유의수준 0.05 기준 통계적 결론
alpha = 0.05

print(f"[통계적 결론]")
print(f"t-통계량 = {t_stat:.4f}")
print(f"p-value  = {p_value_t:.5f}")
print(f"alpha    = {alpha}")

if p_value_t < alpha:
    conclusion = (
        "귀무가설(H0)을 기각한다.\n"
        "  -> 통계적으로 유의한 수준에서, 과자 봉지의 실제 평균 무게는\n"
        "     목표 중량 150g과 다르다고 판단할 수 있다.\n"
        f"  -> 표본평균({sample_weights.mean():.2f}g)이 150g보다 높으므로,\n"
        "     실제로는 목표보다 더 무겁게(과다 충전) 생산되고 있을 가능성이 있다."
    )
else:
    conclusion = (
        "귀무가설(H0)을 기각할 수 없다.\n"
        "  -> 표본 결과만으로는 실제 평균 무게가 150g과 다르다고 볼 만한\n"
        "     충분한 통계적 근거가 없다.\n"
        "  -> 즉, 표본평균의 차이는 우연(표본 변동)의 범위 내에 있을 수 있다."
    )

print(conclusion)
```

```python
# [문제 2] Q4. H₀가 사실(평균 = 150g)이라고 가정하고, 30개짜리 표본을 10,000번 뽑아
# 각 표본의 평균을 리스트에 저장해봅시다.
# 힌트: 표준편차는 우리가 가진 표본의 표준편차를 사용한다고 가정하고, for 반복문을 사용해보세요.

# 여기에 코드를 작성해주세요.

# [문제 2] Q4. H0가 사실이라는 가정 하의 시뮬레이션

# H0: 실제 평균 무게 = 150g (목표 중량)
# 표준편차는 우리가 가진 표본의 표준편차(sample_std)를 그대로 사용한다고 가정하고,
# "30개짜리 표본을 뽑는 실험"을 10,000번 반복해서 각 표본의 평균을 저장한다.

np.random.seed(42)

sample_size = 30
n_simulations = 10000
sample_std = sample_weights.std(ddof=1)  # 우리가 가진 표본의 표준편차 사용

simulated_means = []  # 각 시뮬레이션의 표본평균을 저장할 리스트

for _ in range(n_simulations):
    # H0가 사실이라는 가정 하에, 평균=150g, 표준편차=sample_std인 정규분포에서
    # 30개짜리 표본을 뽑는다.
    fake_sample = np.random.normal(loc=pop_mean, scale=sample_std, size=sample_size)
    simulated_means.append(fake_sample.mean())

simulated_means = np.array(simulated_means)
```

# 문제 3. 어떤 온라인 학습 방식이 더 효과적일까? (난이도: 중)

**📘 문제**

- 한 교육 기업에서 두 가지 다른 온라인 학습 방식(A, B)을 개발했습니다.

- 방식 A가 방식 B보다 학생들의 성적 향상에 더 효과적인지 알아보기 위해,
  두 그룹의 학생들에게 각각 다른 방식으로 한 달간 학습시킨 후 시험을 보게 했습니다.

- 이번 실습에서는 **이표본 t-검정(Two-sample t-Test)**으로 두 방식의 평균 점수 차이를 검증합니다.
  단, t-검정의 기본 가정인 **정규성**과 **등분산성**을 먼저 확인해야 합니다.

- 또한 **"두 방식의 효과가 완전히 똑같다면"** 우연히 나타날 수 있는 점수 차이의 범위를 시뮬레이션으로 확인해 봅니다.

```python
# 데이터 생성 (두 학습 방식의 시험 점수)
np.random.seed(0)
group_a_scores = np.random.normal(loc=85, scale=8, size=50)
group_b_scores = np.random.normal(loc=80, scale=7, size=50)

df_scores = pd.DataFrame({'점수': np.concatenate([group_a_scores, group_b_scores]),
                          '학습방식': ['A']*50 + ['B']*50})

# 두 그룹의 점수 분포 시각화
sns.boxplot(x='학습방식', y='점수', data=df_scores)
plt.title("학습 방식에 따른 시험 점수 분포")
plt.show()
```

**📌 아래를 수행해 보세요:**

- 두 그룹에 대해 **정규성 검정(Shapiro-Wilk test)**을 수행해 봅시다. (H₀: 데이터는 정규분포를 따른다.)

- **등분산성 검정(Levene's test)**을 수행해 봅시다. (H₀: 두 그룹의 분산은 같다.)

- 가정 검토 결과에 따라 `equal_var` 인수를 설정하여 **이표본 t-검정**을 수행해 봅시다.

- 두 그룹 데이터를 모두 합친 뒤 무작위로 다시 나누는 방식으로, **"두 방식의 효과가 같을 때"**의 평균 차이 분포를 시뮬레이션해 봅시다.

```python
# [문제 3] Q1. 두 그룹에 대해 각각 정규성 검정(Shapiro-Wilk test)을 수행하고 p-값을 출력해봅시다.
# H₀: 데이터는 정규분포를 따른다.

# 여기에 코드를 작성해주세요.
# [문제 3] Q1. 정규성 검정 (Shapiro-Wilk test)

# H0: 데이터는 정규분포를 따른다.
# H1: 데이터는 정규분포를 따르지 않는다.
# (정규성 검정은 뒤에 나올 t-검정을 쓸 자격이 있는지 확인하는 사전 점검 단계)

stat_a, p_a = stats.shapiro(group_a_scores)
stat_b, p_b = stats.shapiro(group_b_scores)

print(f"그룹 A: 통계량={stat_a:.4f}, p-value={p_a:.4f}")
print(f"그룹 B: 통계량={stat_b:.4f}, p-value={p_b:.4f}")

#=> 그룹 A: p-value(0.8766) >= 0.05 → H0 기각 못함, 정규분포를 따른다고 볼 수 있음
#=> 그룹 B: p-value(0.8366) >= 0.05 → H0 기각 못함, 정규분포를 따른다고 볼 수 있음
```

```python
# [문제 3] Q2. 두 그룹에 대해 등분산성 검정(Levene's test)을 수행하고 p-값을 출력해봅시다.
# H₀: 두 그룹의 분산은 같다.

# 여기에 코드를 작성해주세요.

# [문제 3] Q2. 등분산성 검정 (Levene's test)

# H0: 두 그룹의 분산은 같다. (등분산)
# H1: 두 그룹의 분산은 다르다. (이분산)
# (등분산성 검정은 t-검정에서 pooled 방식을 쓸지, Welch 방식을 쓸지를 결정하는 사전 점검 단계)

stat_lev, p_lev = stats.levene(group_a_scores, group_b_scores)

print(f"통계량: {stat_lev:.4f}")
print(f"p-value: {p_lev:.4f}")
```

```python
# [문제 3] Q3. 이표본 t-검정을 수행하여 t-통계량과 p-값을 구해봅시다.
# H₀: 두 그룹의 평균은 같다. / H₁: 두 그룹의 평균은 다르다.
# 등분산성 가정을 만족했는지 여부에 따라 equal_var 인수를 설정해보세요.

# 여기에 코드를 작성해주세요.

# [문제 3] Q3. 이표본 t-검정 (Independent two-sample t-test)

# H0: 두 그룹(A, B)의 평균은 같다.
# H1: 두 그룹(A, B)의 평균은 다르다. (양측검정)

# 앞서 Levene 검정에서 등분산성 가정이 기각되었으므로(p=0.0150 < 0.05),
# equal_var=False 로 설정하여 Welch's t-test를 수행한다.

use_equal_var = (p_lev >= alpha)  # 등분산성 기각 못했을 때만 True

t_stat2, p_value2 = stats.ttest_ind(group_a_scores, group_b_scores, equal_var=use_equal_var)

print(f"t-통계량: {t_stat2:.4f}")
print(f"p-value: {p_value2:.5f}")
```

```python
# [문제 3] Q4. H₀가 사실(두 그룹의 평균이 같다)이라고 가정하고 시뮬레이션을 10,000번 수행해봅시다.
# 두 그룹의 데이터를 모두 합친 뒤(np.concatenate), 매 반복마다 비복원추출로 50개(가상 A그룹)와
# 나머지 50개(가상 B그룹)로 나누어 두 그룹의 평균 차이를 계산하고 리스트에 저장해보세요.
# 힌트: np.random.permutation 을 사용해보세요.

# 여기에 코드를 작성해주세요.

# [문제 3] Q4. H0가 사실이라는 가정 하의 순열검정(Permutation Test) 시뮬레이션

# H0: 두 그룹의 평균은 같다.
# -> "학습방식 A/B라는 라벨 자체가 의미 없다"는 뜻이므로, 두 그룹을 하나로 합친 뒤
#    라벨을 무작위로 다시 섞어서(비복원추출) 50개/50개로 나누는 걸 10,000번 반복하면,
#    "우연히 라벨을 나눴을 때 평균 차이가 얼마나 벌어질 수 있는지"의 분포를 만들 수 있다.

np.random.seed(0)

combined = np.concatenate([group_a_scores, group_b_scores])
n_a = len(group_a_scores)  # 50
n_total = len(combined)    # 100
n_simulations = 10000

simulated_diffs = []  # 각 시뮬레이션의 (가상 A평균 - 가상 B평균)을 저장할 리스트

for _ in range(n_simulations):
    shuffled = np.random.permutation(combined)   # 전체 100개를 무작위로 섞음
    fake_a = shuffled[:n_a]      # 앞 50개 -> 가상 A그룹
    fake_b = shuffled[n_a:]      # 나머지 50개 -> 가상 B그룹
    diff = fake_a.mean() - fake_b.mean()
    simulated_diffs.append(diff)

simulated_diffs = np.array(simulated_diffs)

#"두 그룹의 평균이 같다"(H0)는 게 사실이라면,
#   ↓
#애초에 'A그룹', 'B그룹'이라는 라벨 구분 자체가 의미가 없다는 뜻
#   ↓
#그렇다면 100개 데이터를 다 섞어서 아무렇게나 50개/50개로 나눠도,
#   "우연히 그렇게 나뉜 A/B"나 마찬가지다
#   ↓
#이걸 10,000번 반복해서 "라벨을 무작위로 섞었을 때 평균 차이가 얼마나 벌어지는지" 분포를 만든다

#이런 방식을 **순열검정(Permutation Test)**이라고 합니다. 정규분포 같은 이론적 가정 없이, 가진 데이터 자체를 재활용해서 "우연"이 어떤 모습인지 직접 만들어보는 방법이라 더 직관적이고 가정에 덜 의존적인 게 장점이에요.
```

```python
# [문제 3] Q5. 시뮬레이션으로 얻은 평균 차이의 분포를 히스토그램으로 그려봅시다.
# 실제 관찰된 평균 차이(A그룹 평균 - B그룹 평균)를 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.
# (안전장치) observed_diff가 아직 없다면 여기서 다시 계산
observed_diff = group_a_scores.mean() - group_b_scores.mean()

# [문제 3] Q5. 시뮬레이션 결과 히스토그램 시각화
plt.figure(figsize=(10, 6))

# 히스토그램: H0(두 그룹 평균 동일)가 사실일 때의 평균 차이 분포
sns.histplot(simulated_diffs, bins=50, kde=True, color='steelblue',
             label='순열검정 시뮬레이션 결과 (H0 가정, n=10,000회)')

# 실제 관찰된 평균 차이를 세로선으로 표시
plt.axvline(x=observed_diff, color='red', linestyle='--', linewidth=2,
            label=f'실제 관찰된 평균 차이 ({observed_diff:.2f})')

plt.title('H0(두 그룹 평균이 같음)가 사실일 때, 평균 차이(A-B)의 분포\n(순열검정, 10,000회 시뮬레이션)', fontsize=13)
plt.xlabel('평균 차이 (A - B)')
plt.ylabel('빈도 (횟수)')
plt.legend()
plt.tight_layout()
plt.savefig('scores_simulation_histogram.png', dpi=120)

# 순열검정 기반 p-value (양측) 계산: 관찰값의 절대값보다 크거나 같은 비율
perm_p_value = np.mean(np.abs(simulated_diffs) >= np.abs(observed_diff))
print(f"순열검정 기반 p-value(양측): {perm_p_value:.5f}")
print(f"(이론적 Welch's t-test p-value: {p_value2:.5f} 와 비교)")
```

```python
# [문제 4] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 연령대별로 장르 차이가 없다
# H₁ (대립가설): 있다

# [문제 4] Q1. 귀무가설(H0)과 대립가설(H1)

# H0 (귀무가설): 연령대와 선호 장르는 서로 관련이 없다. (독립이다)
#               즉, 어느 연령대든 액션/로맨스를 선호하는 비율은 비슷하다.

# H1 (대립가설): 연령대와 선호 장르는 서로 관련이 있다. (독립이 아니다)
#               즉, 연령대에 따라 선호하는 장르 비율이 다르다.

# (참고) 지금까지와 마찬가지로 H0은 "특별한 관계가 없다"는 심심한 기본 가정이고,
#        H1이 우리가 데이터로 확인하고 싶은 "관련이 있다"는 흥미로운 주장입니다.
```

```python
# [문제 4] Q2. 카이제곱 독립성 검정을 수행하고 카이제곱 통계량, p-값, 자유도, 기대 빈도를 구해봅시다.
# 기대 빈도는 관측 빈도와 비교하기 쉽도록 데이터프레임 형태로 출력해보세요.

# 여기에 코드를 작성해주세요.
# [문제 4] Q2. 카이제곱 독립성 검정 (Chi-square test of independence)
chi2_stat, p_value_chi2, dof, expected = stats.chi2_contingency(observed)

# 기대 빈도를 관측 빈도와 같은 형태의 데이터프레임으로 변환
expected_df = pd.DataFrame(expected, index=observed.index, columns=observed.columns)

print(f"카이제곱 통계량: {chi2_stat:.4f}")
print(f"p-value: {p_value_chi2:.5f}")
print(f"자유도(dof): {dof}")
print(expected_df.round(2))
```

```python
# [문제 4] Q3. H₀가 사실(두 변수가 독립)이라고 가정하고 시뮬레이션을 5,000번 수행해봅시다.
# 각 반복마다 300명의 가상 고객에게 연령대와 장르를 '독립적으로' 할당한 후,
# 가상 분할표를 만들고 카이제곱 통계량을 계산하여 리스트에 저장해보세요.
# 힌트 1: 전체 비율(observed.sum(axis=0) / 전체, observed.sum(axis=1) / 전체)을 확률로 사용
# 힌트 2: np.random.choice 로 300명의 연령대와 장르를 각각 독립적으로 생성
# 힌트 3: pd.crosstab 으로 분할표 생성 후 stats.chi2_contingency 로 통계량 계산

# 여기에 코드를 작성해주세요.

# [문제 4] Q3. H0가 사실이라는 가정 하의 시뮬레이션

# H0: 연령대와 선호 장르는 서로 독립이다.
# -> 전체 데이터에서 관찰된 "연령대 비율"과 "장르 비율"만 그대로 유지한 채,
#    두 변수를 서로 완전히 무관하게(독립적으로) 무작위 배정하면
#    "우연히 독립적으로 나온 카이제곱 통계량"의 분포를 만들 수 있다.

np.random.seed(42)

n_total = observed.values.sum()  # 300명

age_labels = observed.index.tolist()        # ['20대', '30대', '40대']
genre_labels = observed.columns.tolist()    # ['액션', '로맨스']

age_probs = (observed.sum(axis=1) / n_total).values     # 연령대별 비율
genre_probs = (observed.sum(axis=0) / n_total).values   # 장르별 비율

n_simulations = 5000
simulated_chi2 = []

for _ in range(n_simulations):
    # H0가 사실이라는 가정 하에, 연령대와 장르를 각각 독립적으로 무작위 배정
    fake_ages = np.random.choice(age_labels, size=n_total, p=age_probs)
    fake_genres = np.random.choice(genre_labels, size=n_total, p=genre_probs)

    # 가상 분할표 생성 (범주가 누락되지 않도록 순서 고정)
    fake_table = pd.crosstab(
        pd.Categorical(fake_ages, categories=age_labels),
        pd.Categorical(fake_genres, categories=genre_labels)
    )

    chi2_sim, _, _, _ = stats.chi2_contingency(fake_table)
    simulated_chi2.append(chi2_sim)

simulated_chi2 = np.array(simulated_chi2)

print(f"[시뮬레이션 설정]")
print(f"반복 횟수: {n_simulations:,}회")
...
print(f"시뮬레이션 카이제곱 통계량 평균: {simulated_chi2.mean():.3f} ...")

#H0(독립)가 정말 사실이라면
#   ↓
#"어느 연령대에 속하는가"와 "어느 장르를 고르는가"는 서로 아무 상관이 없어야 한다
#   ↓
#그래서 두 변수를 "각자 따로" 무작위로 뽑아서 붙여도(np.random.choice를 두 번 독립적으로 호출),
#H0이 맞는 세상을 그대로 재현하는 셈이 된다

#여기서 중요한 포인트는, 연령대와 장르를 각각 별도로 np.random.choice를 호출했다는 겁니다. 만약 둘을 연결지어서 뽑았다면(예: 20대는 액션을 더 많이 뽑도록) 그건 이미 "관련이 있다"는 걸 가정하는 셈이라 H0 시뮬레이션이 아니게 됩니다. 완전히 독립적으로 뽑아야 "우연히 아무 관계 없을 때 카이제곱 통계량이 얼마나 나오는지"를 제대로 재현할 수 있어요.
```

```python
# [문제 4] 데이터를 어떻게 읽을까요?
# 여기에 의견을 작성해주세요.

기대 빈도의 의미: "만약 연령대와 장르가 정말 아무 관련이 없다면(H0이 맞다면), 이론적으로 나왔어야 할 숫자"입니다.

기대 빈도 = (해당 행의 합계 × 해당 열의 합계) / 전체 합계

예를 들어 20대·액션 칸의 기대 빈도(50)는, "전체 응답자 중 액션 선호 비율이 50%이니, 만약 연령대와 무관하다면 20대 100명 중에서도 딱 50명이 액션을 골랐어야 한다"는 계산입니다.

차이가 클수록 카이제곱 통계량은 커집니다.

카이제곱 통계량 = Σ [(관측 빈도 - 기대 빈도)² / 기대 빈도]

이 공식 자체가 "관측값과 기대값의 차이를 제곱해서 다 더한 것"이라, 차이가 벌어질수록 통계량이 커지도록 설계되어 있습니다. 우리 데이터에서 20대(관측 70 vs 기대 50)와 40대(관측 30 vs 기대 45)처럼 차이가 큰 칸들이 있었기 때문에 카이제곱 통계량이 26.91이라는 큰 값으로 나온 것입니다.

    2.시각화한 것: "연령대와 장르가 정말 서로 독립이라면(H0이 맞다면), 300명에게 두 변수를 완전히 무관하게 무작위로 배정했을 때 카이제곱 통계량이 어떻게 분포하는지"를 5,000번 시뮬레이션으로 보여준 것입니다.

관찰값은 극단적입니다. 히스토그램 대부분은 0~8 사이에 몰려 있고(평균 2.026), 실제 관찰된 카이제곱 통계량(26.91)은 이 분포의 몸통에서 완전히 벗어난 오른쪽 끝자락에 위치합니다. 5,000번 중 단 한 번도 이만큼 큰 값이 나오지 않았을 정도로 드문 값입니다.

    3.p-value(약 0.0000014) < 유의수준(0.05)
   ↓
H0(독립) 기각
   ↓
연령대와 선호 장르는 통계적으로 유의하게 관련이 있다

이유는 명확합니다: "연령대와 장르가 무관하다"는 가정 하에서는 우리가 실제로 관찰한 정도의 큰 차이(카이제곱=26.91)가 나올 확률이 700만 분의 1 수준으로 극히 낮습니다. 이 정도로 희귀한 결과가 실제로 나왔다는 건, 애초에 "무관하다"는 가정 자체가 틀렸다고 보는 게 훨씬 합리적입니다.

4.핵심 패턴: 20대는 액션(70명)을 로맨스(30명)보다 확실히 선호, 40대는 반대로 로맨스(60명)를 액션(30명)보다 확실히 선호. 30대는 중간(50 vs 60)이지만 로맨스 쪽으로 살짝 기움.

활용 가능한 전략

연령대별 타겟 상영관 편성: 대학가나 젊은 층 유동인구가 많은 지점은 액션/블록버스터 상영 비중을 높이고, 중장년층 거주 지역 지점은 로맨스/드라마 비중을 높이는 식으로 지점별 상영 스케줄 차별화가 가능합니다.
연령 타겟 프로모션: 20대 대상 액션 영화 개봉 시 대학생 할인이나 심야 상영 프로모션을, 40대 대상 로맨스 영화 개봉 시 평일 오후·시니어 할인 등을 연계하면 마케팅 효율을 높일 수 있습니다.
큐레이션/추천 시스템: 영화관 앱이나 멤버십에서 연령대 정보를 활용해 "회원님 연령대에서 인기 있는 영화"를 추천하는 개인화 기능에 활용할 수 있습니다.
신작 배급 전략: 신규 영화 개봉 시, 장르에 따라 어느 연령대를 주 타겟으로 광고를 집중할지 결정하는 데 참고할 수 있습니다 (예: 신작 액션 영화 → 20대 타겟 SNS 광고 집중).

주의할 점

이 데이터는 연령대와 장르 두 변수 사이의 "관련성"만 보여줄 뿐, 왜 그런 차이가 나는지(취향 형성 원인)까지 설명해주지는 않습니다.
표본이 300명으로 한정되어 있고, 특정 시기·특정 지역의 데이터일 수 있으니, 실제 정책에 반영하기 전에는 더 큰 표본과 여러 시점의 데이터로 재검증하는 것이 안전합니다.
30대처럼 뚜렷하지 않은 경계층에 대해서는 지나치게 단정적인 전략보다는 두 장르를 균형 있게 제공하는 접근이 더 적절할 수 있습니다.
```

# 문제 5. 미니 프로젝트 - 어떤 신규 비료가 가장 효과적일까? (난이도: 상)

**📘 문제**

- 한 농업 연구소에서 새로 개발한 비료 3종류(A, B, C)의 생산량 증대 효과를 비교하고자 합니다.

- 동일한 조건의 밭 30개를 준비하여, 각각 10개씩 비료 A, B, C를 투여한 후 수확량을 측정했습니다.

- 이번 실습에서는 **분산분석(ANOVA)**으로 세 비료 간 평균 수확량에 차이가 있는지 검증하고,
  차이가 있다면 **사후분석(Tukey's HSD)**으로 어떤 비료끼리 차이가 나는지 확인합니다.

- 마지막으로 **"세 비료의 효과가 완전히 똑같다면"** 어떤 F-통계량이 나올 수 있는지 시뮬레이션하고,
  분석 결과를 바탕으로 연구소에 어떤 비료를 추천할지까지 생각해 봅니다.

```python
# 데이터 생성 (비료 종류별 수확량)
np.random.seed(123)
fertilizer_a = np.random.normal(loc=105, scale=10, size=10)
fertilizer_b = np.random.normal(loc=120, scale=12, size=10)
fertilizer_c = np.random.normal(loc=108, scale=9, size=10)

# 데이터프레임으로 변환 (사후분석을 위해)
df = pd.DataFrame({'수확량': np.concatenate([fertilizer_a, fertilizer_b, fertilizer_c]),
                   '비료종류': ['A']*10 + ['B']*10 + ['C']*10})

# 데이터 탐색 시각화
sns.boxplot(x='비료종류', y='수확량', data=df)
plt.title("비료 종류에 따른 수확량 분포")
plt.show()
```

**📌 아래를 수행해 보세요:**

- 세 그룹에 대해 분산분석(ANOVA)을 수행하여 F-통계량과 p-값을 구해봅시다.

- ANOVA 결과가 유의미할 경우에만 **Tukey's HSD 사후분석**을 수행하고 결과를 해석해 봅시다.

- **"세 비료의 효과가 완전히 똑같다면"** 어떤 F-통계량들이 나오는지 5,000번 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 실제 관찰된 F-통계량이 얼마나 극단적인 값인지 확인해 봅시다.

- 분석 결과를 종합하여 연구소에 추천할 비료를 정해 봅시다.

```python
# [문제 5] Q1. 분산분석(ANOVA)을 수행하여 F-통계량과 p-값을 구해봅시다.
# H₀: 세 비료의 평균 수확량은 모두 같다.
# H₁: 적어도 하나 이상의 비료 평균 수확량은 다르다.

# 여기에 코드를 작성해주세요.

# [문제 5] Q1. 분산분석(ANOVA)

# H0: 세 비료의 평균 수확량은 모두 같다. (mu_A = mu_B = mu_C)
# H1: 적어도 하나 이상의 비료 평균 수확량은 다르다.

from scipy import stats

f_stat, p_value_anova = stats.f_oneway(fertilizer_a, fertilizer_b, fertilizer_c)

print(f"F-통계량: {f_stat:.4f}")
print(f"p-value: {p_value_anova:.5f}")
```

```python
# [문제 5] Q2. ANOVA의 p-값이 유의수준 0.05보다 작을 경우에만
# Tukey's HSD 사후분석을 수행하고 결과를 출력해봅시다.
# 힌트: pairwise_tukeyhsd(endog=수확량, groups=비료종류, alpha=0.05)

# 여기에 코드를 작성해주세요.

# [문제 5] Q2. Tukey's HSD 사후분석

# ANOVA에서 유의한 차이가 발견된 경우(p < alpha)에만,
# 구체적으로 "어느 비료 쌍끼리" 차이가 나는지 사후분석으로 확인한다.

from statsmodels.stats.multicomp import pairwise_tukeyhsd

if p_value_anova < alpha:
    print("[Tukey's HSD 사후분석 결과]")
    print("(ANOVA에서 유의한 차이가 발견되었으므로 사후분석을 수행합니다)")

    tukey_result = pairwise_tukeyhsd(endog=df['수확량'], groups=df['비료종류'], alpha=0.05)
    print(tukey_result)
else:
    print(f"ANOVA 결과 p-value({p_value_anova:.5f})가 유의수준({alpha})보다 크므로,")
    print("사후분석(Tukey's HSD)을 수행하지 않습니다.")

#A와 B는 유의하게 다릅니다 (B가 A보다 약 23.7 더 높음)
#B와 C도 유의하게 다릅니다 (B가 C보다 약 18.8 더 높음)
#A와 C는 유의한 차이가 없습니다 (차이 4.84는 우연 범위 안)
#결론: 비료 B는 A, C보다 확실히 수확량이 높고, A와 C는 비슷한 수준입니다.
```

```python
# [문제 5] Q3. H₀가 사실(세 그룹의 평균이 모두 같다)이라고 가정하고 시뮬레이션을 5,000번 수행해봅시다.
# 모든 데이터를 합친 평균과 표준편차를 가진 H₀ 모집단에서 10개씩 3개의 가상 그룹을 생성하고,
# 이 가상 그룹들로 ANOVA를 수행하여 F-통계량을 리스트에 저장해보세요.

# 여기에 코드를 작성해주세요.

# [문제 5] Q3. H0가 사실이라는 가정 하의 시뮬레이션

# H0: 세 비료의 평균 수확량은 모두 같다.
# -> 전체 데이터(30개)를 하나의 모집단으로 보고, 그 평균/표준편차를 가진
#    가상의 모집단에서 10개씩 3그룹을 뽑아 ANOVA를 수행하는 걸 5,000번 반복한다.

np.random.seed(123)

all_data_mean = df['수확량'].mean()
all_data_std = df['수확량'].std(ddof=1)
group_size = 10
n_simulations = 5000

simulated_f_stats = []  # 각 시뮬레이션의 F-통계량을 저장할 리스트

for _ in range(n_simulations):
    # H0가 사실이라는 가정 하에, 세 그룹 모두 같은 분포(평균, 표준편차)에서 뽑는다.
    group1 = np.random.normal(loc=all_data_mean, scale=all_data_std, size=group_size)
    group2 = np.random.normal(loc=all_data_mean, scale=all_data_std, size=group_size)
    group3 = np.random.normal(loc=all_data_mean, scale=all_data_std, size=group_size)

    f_sim, _ = stats.f_oneway(group1, group2, group3)
    simulated_f_stats.append(f_sim)

simulated_f_stats = np.array(simulated_f_stats)

#전체 데이터 평균: 111.811, 전체 데이터 표준편차: 15.769
#반복 횟수: 5,000회
#각 실험: 위 분포에서 10개씩 3개 그룹을 뽑아 ANOVA 수행
#시뮬레이션 F-통계량 예시(앞 5개): [1.162 1.969 0.919 0.017 0.19 ]
#시뮬레이션 F-통계량 평균: 1.043 (이론적으로 H0 하에서 F의 기댓값은 1에 가까움)

#여기서 "H0가 사실"이라는 가정은 **"세 그룹이 전부 같은 하나의 모집단에서 나온 것"**이라는 뜻입니다. 그래서 실제 A, B, C 각각의 평균/표준편차를 쓰는 게 아니라, **30개 전체를 하나로 합친 평균(111.811)과 표준편차(15.769)**를 가진 단일 분포를 만들고, 거기서 세 그룹을 똑같이 뽑는 겁니다.
```

- ANOVA 검정의 p-값을 통해 어떤 결론을 내릴 수 있나요? 이 결과는 "세 비료가 모두 동일한 효과를 가진다"는 뜻인가요, "적어도 하나는 다르다"는 뜻인가요?

- `meandiff` 열의 값은 어떤 정보를 주나요? 신뢰구간(`lower`, `upper`)이 0을 포함하는지 여부는 무엇을 뜻할까요?

- 그룹별 표본 크기가 10개로 작다는 점은 이 결론의 신뢰도에 어떤 영향을 줄까요? 어떻게 개선할 수 있을까요?

```python
# [문제 5] 데이터를 어떻게 읽을까요?
# 여기에 의견을 작성해주세요.

적어도 하나는 다르다

reject=True는 "해당 두 그룹의 귀무가설(이 둘의 평균은 같다)을 기각한다", 즉 **"이 두 그룹 간에는 통계적으로 유의한 차이가 있다"**는 뜻입니다.
      A      B  23.6754 0.0006   True   ← A와 B는 유의하게 다르다
     A      C   4.8429 0.6569   False  ← A와 C는 차이가 없다고 봐도 됨
     B      C -18.8325 0.0054   True   ← B와 C는 유의하게 다르다
True인 조합(A-B, B-C) 은 "비료 B가 A, C와는 확실히 다른 효과를 낸다"
False인 조합(A-C)**은 "A와 C는 사실상 비슷한 효과"

meandiff: 두 그룹 평균의 차이(group2 - group1)입니다. 예를 들어 A-B의 meandiff=23.6754는 "B의 평균이 A보다 23.68만큼 높다"는 뜻입니다.
A-B: lower=10.04, upper=37.31  → 구간이 0을 포함하지 않음 (전부 양수) → 유의함 (True)
A-C: lower=-8.79, upper=18.48  → 구간이 0을 포함함 (음수~양수 걸쳐있음) → 유의하지 않음 (False)
신뢰구간이 0을 포함하지 않는다 = "이 두 그룹의 진짜 평균 차이는 0(차이 없음)일 가능성이 낮다" → 실제 차이가 있다고 볼 근거
신뢰구간이 0을 포함한다 = "차이가 0(즉, 차이가 없음)일 수도 있다"는 걸 배제하지 못함 → 유의한 차이라고 말하기 어려움
즉, reject 열의 True/False는 사실 이 신뢰구간이 0을 포함하는지를 자동으로 판단해서 표시해주는 것뿐이라, 신뢰구간을 직접 보면 왜 True/False가 나왔는지 그 근거를 눈으로 확인할 수 있습니다.

          표본이 작으면 표준오차가 커지고, 이는 신뢰구간이 넓어진다는 뜻입니다. 신뢰구간이 넓으면 "0을 포함하느냐 안 하느냐"의 경계가 애매해지기 쉽고, 실제로는 차이가 있는데도 표본이 작아서 못 잡아내는 경우(2종 오류, 검정력 부족)가 생길 수 있습니다.

표본 크기 늘리기: 그룹당 10개가 아니라 30개, 50개로 늘리면 표준오차가 줄어들어(표준오차 ∝ 1/√n) 신뢰구간이 좁아지고, 더 미세한 차이도 탐지할 수 있는 검정력이 높아집니다.
사전 검정력 분석(Power Analysis): 실험을 설계하기 전에 "얼마나 큰 차이를 탐지하고 싶은가"를 정하고, 그에 필요한 최소 표본 크기를 미리 계산해서 실험을 설계하는 것이 이상적입니다.
반복 실험(replication): 같은 실험을 여러 시즌/여러 밭 조건에서 반복해서, 이번 결과가 우연한 환경 요인(날씨, 토양 등) 때문이 아닌지 재확인하는 것도 신뢰도를 높이는 방법입니다.
```
